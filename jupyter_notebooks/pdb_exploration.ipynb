```python
import tarfile



def list_all_unique_elements(tar_gz_path):
    unique_elements = set()

    with tarfile.open(tar_gz_path, "r:gz") as tar:
        for member in tar.getmembers():
            if member.isfile() and member.name.endswith(".pdb"):
                with tar.extractfile(member) as file:
                    for line in file:
                        if line.startswith(b"ATOM") or line.startswith(b"HETATM"):
                            atom_name = line[12:16].strip().decode("utf-8")
                            unique_elements.add(atom_name)

    return unique_elements
```

```python
tar_gz_path = "/dataset/downloads/sep_prot_frames_DPP8_12i_correct_replica1.tar.gz"
unique_atoms = list_all_unique_elements(tar_gz_path)
print(unique_atoms)
```
