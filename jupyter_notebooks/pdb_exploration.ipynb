```python
import tarfile

def list_header_lines(tar_gz_path, num_lines=10):
    with tarfile.open(tar_gz_path, "r:gz") as tar:
        # get members ending with .pdb
        for member in tar.getmembers():
            if member.isfile() and member.name.endswith(".pdb"):
                member = tar.getmember(member.name) 
                break       
        with tar.extractfile(member) as file:
            for i in range(num_lines):
                print(file.readline().decode("utf-8").strip())

def list_all_unique_elements(tar_gz_path):
    unique_elements = set()

    with tarfile.open(tar_gz_path, "r:gz") as tar:
        # get members ending with .pdb
        for member in tar.getmembers():
            if member.isfile() and member.name.endswith(".pdb"):
                member = tar.getmember(member.name)        
                break
        with tar.extractfile(member) as file:
            for line in file:
                if line.startswith(b"ATOM") or line.startswith(b"HETATM"):
                    atom_name = line[12:16].strip().decode("utf-8")
                    unique_elements.add(atom_name)

    return unique_elements

def list_indexes_of_elements(tar_gz_path, element):
    indices = []

    with tarfile.open(tar_gz_path, "r:gz") as tar:
        # get members ending with .pdb
        for member in tar.getmembers():
            if member.isfile() and member.name.endswith(".pdb"):
                member = tar.getmember(member.name)      
                break  
        with tar.extractfile(member) as file:
            for line in file:
                if line.startswith(b"ATOM") or line.startswith(b"HETATM"):
                    atom_name = line[12:16].strip().decode("utf-8")
                    if atom_name == element:
                        index = int(line[6:11].strip())
                        indices.append(index)
    return indices
```

```python
tar_gz_path = "/project_antwerp/dataset/downloads/sep_prot_frames_DPP8_12i_correct_replica1.tar.gz"
unique_atoms = list_all_unique_elements(tar_gz_path)
print(unique_atoms)
```

```python
# print all indexes of CA atoms
ca_indices = list_indexes_of_elements(tar_gz_path, "CA")
print(ca_indices)
```
