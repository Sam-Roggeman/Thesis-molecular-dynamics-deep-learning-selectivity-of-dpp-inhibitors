```python
percent_str = str(int(15))
cache_folder = os.environ.get("HF_CACHE_DIR")
hf_token = os.environ.get("HF_TOKEN")
# set size in string format
downloaded_cache_folder = os.path.join(cache_folder, "downloaded_cache")
train_dataloader = datasets.load_dataset("Sam-Roggeman/SamRoggeman_Thesis_Dataset", split=f"train[:15%]", token=hf_token, num_proc=1, cache_dir=downloaded_cache_folder)
```

```python
# view the classes in the dataset
print(train_dataloader.features)
# view the first example
# view the labels in the dataset
labels = train_dataloader.unique("binding_type")
print(labels)
# print nr of each label
label_counts = count = {label: 0 for label in labels}
# iterate through the dataset and count the labels
for example in train_dataloader:
    label_counts[example["binding_type"]] += 1


print(label_counts)
```
