```python
dataset_folder = "../data/dataset/full_dataset/"
ds = datasets.load_from_disk(os.path.join(dataset_folder, "train"))
print(len(ds))
dataset_size = 0.10

ds = get_subset(ds, dataset_size)
len(ds)
```

```python
def count_ligand_names(ds):
    from collections import Counter
    ligand_name_counts = Counter()
    total_processed = 0
    batch_size = 1000
    batch_counter = 0
    total_batches = len(ds) // batch_size
    for i in range(0, len(ds), batch_size):  # process 1000 at a time
        batch = ds[i:i+batch_size]['ligand_name']
        ligand_name_counts.update(batch)
        batch_counter += 1
        print(f"{batch_counter}/{total_batches} batches processed", end='\r')
    return ligand_name_counts
```

```python
count_ligand_names(ds)
```
