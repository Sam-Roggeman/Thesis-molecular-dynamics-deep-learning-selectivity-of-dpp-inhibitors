```python
def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)
```

```python


# load the model and print the number of parameters
import torch
from src.Models.SCNN import SCNN
model = SCNN()
print(f"Number of trainable parameters in SimpleCNN: {count_parameters(model)}")

```

```python
from src.Models.DCNN import DCNN
model = DCNN()
print(f"Number of trainable parameters in CustomDenseNet: {count_parameters(model)}")
```
