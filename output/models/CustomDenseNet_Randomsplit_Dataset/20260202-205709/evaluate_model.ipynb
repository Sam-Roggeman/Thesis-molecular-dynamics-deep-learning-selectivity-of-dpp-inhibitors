```python
import torch
import torch.nn as nn
from dotenv import load_dotenv

# load CUstomDensenet model
from src.Models.DCNN import CustomDenseNet
from src.model_training.metric_functions import all_statistics
import datasets

from src.utils.training_config import TrainingConfig
from src.utils.training_setup import load_and_prepare_test
load_dotenv()
from collections import Counter
```

```python
model = CustomDenseNet(num_classes=5)
# load model weights
model_path = "./CustomDenseNet_Randomsplit_Dataset.pth"
model.load_state_dict(torch.load(model_path))
```

```python
config = TrainingConfig(
    dataset_location="Sam-Roggeman/SamRoggeman_Thesis_Dataset",
    batch_size=32
)
test_loader = load_and_prepare_test(
    config=config
)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
criterion = nn.CrossEntropyLoss()
stats = all_statistics(model=model, dataloader=test_loader, device=device, criterion=criterion)
```

```python
count = Counter(label for batch in test_loader for label in batch["labels"].tolist())
print(count)
```

```python
confusion_matrix = stats["confusion_matrix"]
print(confusion_matrix)
```
